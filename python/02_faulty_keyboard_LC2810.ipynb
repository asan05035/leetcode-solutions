## 2810. Faulty Keyboard
Your laptop keyboard is faulty, and whenever you type a character 'i' on it, it reverses the string that you have written. Typing other characters works as expected.

You are given a 0-indexed string s, and you type each character of s using your faulty keyboard.

Return the final string that will be present on your laptop screen.

### way_1: using for + two pointers in swap

```python
class Solution:
    def finalString(self, s: str) -> str:
        #list_s = list(s)
        ans = list()

        for i in range(0, len(s)):
            if s[i] != 'i':
                ans.append(s[i])
            else:
                '''
                    Reverse a list
                    .reverse() method
                '''
                l = 0
                r = len(ans) - 1

                while l < r:
                    ans[l], ans[r] = ans[r], ans[l]

                    l, r = l + 1, r - 1
                
        #print(ans)

        return ''.join(ans) 
```

### Way_2: Deque() using queue
Reference for deque :[Definition by indently(Youtube)](https://www.youtube.com/watch?v=34SNQHapJYE)

```python
from collections import deque
class Solution:
    def finalString(self, s: str) -> str:
        dq = deque()
        
        flag = False #Used to shift perspective

        for ch in s:
            if ch == 'i':
                flag = not flag
            elif flag:
                dq.appendleft(ch)
            else:
                dq.append(ch)


        #If the flagg is already in right prespective no need to reverse
        if flag:
            dq.reverse()
        
        return ''.join(dq)
```
